```python
# Load CIFAR-10 dataset

(x_train, y_train), (x_test, y_test) = tf.keras.datasets.cifar10.load_data()

print("Dataset loaded successfully!")
print()
print("Training images:", x_train.shape)
print("Training labels:", y_train.shape)
print("Testing images:", x_test.shape)
print("Testing labels:", y_test.shape)
```

```python
# EXPERIMENT 3
# TASK B: CNN MODEL IMPLEMENTATION

import tensorflow as tf
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D
from tensorflow.keras.layers import Flatten, Dense

print("TASK B - CNN IMPLEMENTATION")

# BUILD CNN MODEL

model = Sequential([

    # Input image: 32 x 32 x 3
    # Convolution Layer 1 + ReLU Activation
    Conv2D(
        filters=32,
        kernel_size=(3, 3),
        activation='relu',
        padding='same',
        input_shape=(32, 32, 3)
    ),

    # Max Pooling Layer 1
    MaxPooling2D(pool_size=(2, 2)),

    # Convolution Layer 2 + ReLU Activation
    Conv2D(
        filters=64,
        kernel_size=(3, 3),
        activation='relu',
        padding='same'
    ),

    # Max Pooling Layer 2
    MaxPooling2D(pool_size=(2, 2)),

    # Convolution Layer 3 + ReLU Activation
    Conv2D(
        filters=128,
        kernel_size=(3, 3),
        activation='relu',
        padding='same'
    ),

    # Max Pooling Layer 3
    MaxPooling2D(pool_size=(2, 2)),

    # Flatten Layer
    # Converts feature maps into a 1D vector
    Flatten(),

    # Fully Connected Layer
    Dense(
        128,
        activation='relu'
    ),

    # Softmax Output Layer
    # CIFAR-10 has 10 classes
    Dense(
        10,
        activation='softmax'
    )
])


# DISPLAY MODEL SUMMARY


model.summary()
```

```python
# COMPILE THE CNN MODEL

model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

print("           MODEL COMPILED SUCCESSFULLY")
print("Optimizer : Adam")
print("Loss      : Sparse Categorical Crossentropy")
print("Metric    : Accuracy")
```

```python
# EXPERIMENT NO: 3
# IMPLEMENTATION OF CONVOLUTIONAL NEURAL NETWORKS (CNNs)
# FOR IMAGE CLASSIFICATION
#
# TASK C: MODEL TRAINING AND EVALUATION
# DATASET: CIFAR-10

import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf

print("              EXPERIMENT NO: 3")
print("       CNN IMAGE CLASSIFICATION")
print("Name     : SWATHI V")
print("Roll No  : YOUR_ROLL_NO")
print("Task     : C - Model Training and Evaluation")
print("Dataset  : CIFAR-10")


# 1. LOAD CIFAR-10 DATASET

(x_train, y_train), (x_test, y_test) = tf.keras.datasets.cifar10.load_data()

print("\nDataset loaded successfully!")

print("Original Training Data :", x_train.shape)
print("Original Testing Data  :", x_test.shape)


# 2. NORMALIZE IMAGES

x_train = x_train.astype('float32') / 255.0
x_test = x_test.astype('float32') / 255.0

print("\nImage normalization completed.")


# 3. CREATE VALIDATION DATASET

validation_size = 5000

x_val = x_train[-validation_size:]
y_val = y_train[-validation_size:]

x_train = x_train[:-validation_size]
y_train = y_train[:-validation_size]

print("\nDataset Split:")
print("Training   :", x_train.shape)
print("Validation :", x_val.shape)
print("Testing    :", x_test.shape)


# 4. BUILD CNN MODEL

model = tf.keras.Sequential([

    # Input Layer
    tf.keras.layers.Input(shape=(32, 32, 3)),

    # Convolution Layer 1 + ReLU
    tf.keras.layers.Conv2D(
        filters=32,
        kernel_size=(3, 3),
        activation='relu',
        padding='same'
    ),

    # Max Pooling Layer 1
    tf.keras.layers.MaxPooling2D(
        pool_size=(2, 2)
    ),

    # Convolution Layer 2 + ReLU
    tf.keras.layers.Conv2D(
        filters=64,
        kernel_size=(3, 3),
        activation='relu',
        padding='same'
    ),

    # Max Pooling Layer 2
    tf.keras.layers.MaxPooling2D(
        pool_size=(2, 2)
    ),

    # Convolution Layer 3 + ReLU
    tf.keras.layers.Conv2D(
        filters=128,
        kernel_size=(3, 3),
        activation='relu',
        padding='same'
    ),

    # Max Pooling Layer 3
    tf.keras.layers.MaxPooling2D(
        pool_size=(2, 2)
    ),

    # Flatten Layer
    tf.keras.layers.Flatten(),

    # Fully Connected Layer
    tf.keras.layers.Dense(
        128,
        activation='relu'
    ),

    # Softmax Output Layer
    tf.keras.layers.Dense(
        10,
        activation='softmax'
    )
])


# 5. COMPILE MODEL

model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

print("\nCNN model compiled successfully!")


# 6. DISPLAY MODEL SUMMARY

print("              CNN MODEL SUMMARY")

model.summary()


# 7. TRAIN THE CNN MODEL

epochs = 15
batch_size = 64

print("              MODEL TRAINING")

history = model.fit(
    x_train,
    y_train,
    epochs=epochs,
    batch_size=batch_size,
    validation_data=(x_val, y_val),
    verbose=1
)


# 8. EVALUATE TRAINING DATA

train_loss, train_accuracy = model.evaluate(
    x_train,
    y_train,
    verbose=0
)


# 9. EVALUATE VALIDATION DATA

val_loss, val_accuracy = model.evaluate(
    x_val,
    y_val,
    verbose=0
)


# 10. EVALUATE TESTING DATA

test_loss, test_accuracy = model.evaluate(
    x_test,
    y_test,
    verbose=0
)


# 11. DISPLAY PERFORMANCE RESULTS

print("              MODEL PERFORMANCE")

print(
    "Training Accuracy   :",
    f"{train_accuracy * 100:.2f}%"
)

print(
    "Validation Accuracy :",
    f"{val_accuracy * 100:.2f}%"
)

print(
    "Testing Accuracy    :",
    f"{test_accuracy * 100:.2f}%"
)

print()

print(
    "Training Loss       :",
    f"{train_loss:.4f}"
)

print(
    "Validation Loss     :",
    f"{val_loss:.4f}"
)

print(
    "Testing Loss        :",
    f"{test_loss:.4f}"
)



# 12. ACCURACY VS EPOCH

plt.figure(figsize=(8, 5))

plt.plot(
    history.history['accuracy'],
    label='Training Accuracy'
)

plt.plot(
    history.history['val_accuracy'],
    label='Validation Accuracy'
)

plt.title('Accuracy vs Epoch')
plt.xlabel('Epoch')
plt.ylabel('Accuracy')
plt.legend()
plt.grid(True)

plt.show()


# 13. LOSS VS EPOCH

plt.figure(figsize=(8, 5))

plt.plot(
    history.history['loss'],
    label='Training Loss'
)

plt.plot(
    history.history['val_loss'],
    label='Validation Loss'
)

plt.title('Loss vs Epoch')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.legend()
plt.grid(True)

plt.show()


# 14. FINAL RESULT



print("Training Accuracy   :", f"{train_accuracy * 100:.2f}%")
print("Validation Accuracy :", f"{val_accuracy * 100:.2f}%")
print("Testing Accuracy    :", f"{test_accuracy * 100:.2f}%")

print()
print("Training Loss       :", f"{train_loss:.4f}")
print("Validation Loss     :", f"{val_loss:.4f}")
print("Testing Loss        :", f"{test_loss:.4f}")

print("\nAccuracy vs Epoch    : Plotted")
print("Loss vs Epoch        : Plotted")

print("\nCNN Model Training and Evaluation Completed Successfully!")
```

```python
# ============================================================
# EXPERIMENT NO: 3
# IMPLEMENTATION OF CNN FOR IMAGE CLASSIFICATION
#
# TASK D: PERFORMANCE ANALYSIS
# DATASET: CIFAR-10
# ============================================================

import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import tensorflow as tf

from sklearn.metrics import confusion_matrix
from sklearn.metrics import classification_report


# ============================================================
# STUDENT DETAILS
# ============================================================

name = "SWATHI V"
roll_no = "YOUR_ROLL_NO"

print("====================================================")
print("              EXPERIMENT NO: 3")
print("          TASK D - PERFORMANCE ANALYSIS")
print("====================================================")
print("Name     :", name)
print("Roll No  :", roll_no)
print("Dataset  : CIFAR-10")
print("====================================================")


# --- BEGIN: Code to ensure model, x_test, y_test are defined and model is trained ---
# This section is copied from Task C (cell gfo02k40-C6D) to make this cell self-contained.

# 1. LOAD CIFAR-10 DATASET
(x_train, y_train), (x_test, y_test) = tf.keras.datasets.cifar10.load_data()

# 2. NORMALIZE IMAGES
x_train = x_train.astype('float32') / 255.0
x_test = x_test.astype('float32') / 255.0

# 3. CREATE VALIDATION DATASET
validation_size = 5000
x_val = x_train[-validation_size:]
y_val = y_train[-validation_size:]
x_train = x_train[:-validation_size]
y_train = y_train[:-validation_size]

# 4. BUILD CNN MODEL
model = tf.keras.Sequential([
    tf.keras.layers.Input(shape=(32, 32, 3)),
    tf.keras.layers.Conv2D(filters=32, kernel_size=(3, 3), activation='relu', padding='same'),
    tf.keras.layers.MaxPooling2D(pool_size=(2, 2)),
    tf.keras.layers.Conv2D(filters=64, kernel_size=(3, 3), activation='relu', padding='same'),
    tf.keras.layers.MaxPooling2D(pool_size=(2, 2)),
    tf.keras.layers.Conv2D(filters=128, kernel_size=(3, 3), activation='relu', padding='same'),
    tf.keras.layers.MaxPooling2D(pool_size=(2, 2)),
    tf.keras.layers.Flatten(),
    tf.keras.layers.Dense(128, activation='relu'),
    tf.keras.layers.Dense(10, activation='softmax')
])

# 5. COMPILE MODEL
model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

# 6. TRAIN THE CNN MODEL (using fewer epochs for demonstration in this combined cell)
# For full training, you should run Task C (cell gfo02k40-C6D) separately.
epochs = 5 # Reduced epochs to keep this combined cell quicker to run
batch_size = 64

history = model.fit(
    x_train,
    y_train,
    epochs=epochs,
    batch_size=batch_size,
    validation_data=(x_val, y_val),
    verbose=0 # Set to 0 to suppress verbose training output in this cell
)

# --- END: Code to ensure model, x_test, y_test are defined and model is trained ---


# ============================================================
# 1. CLASS NAMES
# ============================================================

class_names = [
    'Airplane',
    'Automobile',
    'Bird',
    'Cat',
    'Deer',
    'Dog',
    'Frog',
    'Horse',
    'Ship',
    'Truck'
]


# ============================================================
# 2. TESTING ACCURACY
# ============================================================

test_loss, test_accuracy = model.evaluate(
    x_test,
    y_test,
    verbose=0
)

print("\n====================================================")
print("             CLASSIFICATION ACCURACY")
print("====================================================")

print("Testing Loss     :", f"{test_loss:.4f}")
print("Testing Accuracy :", f"{test_accuracy * 100:.2f}%")

print("====================================================")


# ============================================================
# 3. MAKE PREDICTIONS
# ============================================================

print("\nGenerating predictions...")

predictions = model.predict(
    x_test,
    verbose=1
)

# Convert probability values into class numbers
predicted_classes = np.argmax(
    predictions,
    axis=1
)

# Convert actual labels into 1D array
actual_classes = y_test.flatten()

print("\nPredictions generated successfully!")


# ============================================================
# 4. CONFUSION MATRIX
# ============================================================

cm = confusion_matrix(
    actual_classes,
    predicted_classes
)

print("\n====================================================")
print("                 CONFUSION MATRIX")
print("====================================================")

print(cm)


# Display confusion matrix as a heatmap
plt.figure(figsize=(10, 8))

sns.heatmap(
    cm,
    annot=True,
    fmt='d',
    cmap='Blues',
    xticklabels=class_names,
    yticklabels=class_names
)

plt.title('Confusion Matrix - CIFAR-10')
plt.xlabel('Predicted Class')
plt.ylabel('Actual Class')

plt.tight_layout()
plt.show()


# ============================================================
# 5. CLASSIFICATION REPORT
# ============================================================

print("\n====================================================")
print("             CLASSIFICATION REPORT")
print("====================================================")

print(
    classification_report(
        actual_classes,
        predicted_classes,
        target_names=class_names
    )
)


# ============================================================
# 6. SAMPLE PREDICTIONS
# ============================================================

print("\n====================================================")
print("               SAMPLE PREDICTIONS")
print("====================================================")

plt.figure(figsize=(12, 8))

for i in range(12):

    plt.subplot(3, 4, i + 1)

    plt.imshow(x_test[i])

    actual_label = class_names[actual_classes[i]]
    predicted_label = class_names[predicted_classes[i]]

    plt.title(
        "Actual: " + actual_label +
        "\nPredicted: " + predicted_label,
        fontsize=9
    )

    plt.axis('off')

plt.tight_layout()
plt.show()


# ============================================================
# 7. FIND MISCLASSIFIED IMAGES
# ============================================================

misclassified_indices = np.where(
    predicted_classes != actual_classes
)[0]

print("\n====================================================")
print("             MISCLASSIFIED IMAGES")
print("====================================================")

print(
    "Total misclassified images :",
    len(misclassified_indices)
)

print(
    "Total correctly classified :",
    len(actual_classes) - len(misclassified_indices)
)


# ============================================================
# 8. DISPLAY MISCLASSIFIED IMAGES
# ============================================================

number_of_images = min(
    12,
    len(misclassified_indices)
)

plt.figure(figsize=(12, 8))

for i in range(number_of_images):

    index = misclassified_indices[i]

    plt.subplot(3, 4, i + 1)

    plt.imshow(x_test[index])

    actual_label = class_names[
        actual_classes[index]
    ]

    predicted_label = class_names[
        predicted_classes[index]
    ]

    plt.title(
        "Actual: " + actual_label +
        "\nPredicted: " + predicted_label,
        fontsize=9
    )

    plt.axis('off')

plt.tight_layout()
plt.show()


# ============================================================
# 9. DISPLAY PREDICTION PROBABILITY
# ============================================================

print("\n====================================================")
print("          SAMPLE PREDICTION PROBABILITIES")
print("====================================================")

for i in range(5):

    predicted_class = predicted_classes[i]

    confidence = predictions[i][predicted_class] * 100

    print(
        f"Image {i+1}: "
        f"Predicted = {class_names[predicted_class]}, "
        f"Confidence = {confidence:.2f}%"
    )


# ============================================================
# 10. PERFORMANCE ANALYSIS
# ============================================================

print("\n====================================================")
print("              PERFORMANCE ANALYSIS")
print("====================================================")

print(
    f"Overall Testing Accuracy: "
    f"{test_accuracy * 100:.2f}%"
)

print(
    f"Correct Predictions: "
    f"{len(actual_classes) - len(misclassified_indices)}"
)

print(
    f"Incorrect Predictions: "
    f"{len(misclassified_indices)}"
)

print("====================================================")


# ============================================================
# 11. TASK D COMPLETION
# ============================================================

print("\n====================================================")
print("             TASK D COMPLETED")
print("====================================================")

print("✓ Classification Accuracy")
print("✓ Confusion Matrix")
print("✓ Sample Predictions")
print("✓ Misclassified Images")
print("✓ Performance Analysis")

print("\nCNN Performance Analysis Completed Successfully!")
print("====================================================")
```
