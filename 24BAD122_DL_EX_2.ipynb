```python

import numpy as np
import matplotlib.pyplot as plt
```

```python
# Sigmoid Activation Function

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

x = np.linspace(-10, 10, 400)
y = sigmoid(x)

plt.figure(figsize=(8, 5))
plt.plot(x, y)
plt.title("Sigmoid Activation Function")
plt.xlabel("Input (x)")
plt.ylabel("Output")
plt.grid(True)
plt.axhline(0, linewidth=0.8)
plt.axvline(0, linewidth=0.8)
plt.show()
```

```python
# Tanh Activation Function

def tanh(x):
    return np.tanh(x)

x = np.linspace(-10, 10, 400)
y = tanh(x)

plt.figure(figsize=(8, 5))
plt.plot(x, y)
plt.title("Tanh Activation Function")
plt.xlabel("Input (x)")
plt.ylabel("Output")
plt.grid(True)
plt.axhline(0, linewidth=0.8)
plt.axvline(0, linewidth=0.8)
plt.show()
```

```python
# ReLU Activation Function

def relu(x):
    return np.maximum(0, x)

x = np.linspace(-10, 10, 400)
y = relu(x)

plt.figure(figsize=(8, 5))
plt.plot(x, y)
plt.title("ReLU Activation Function")
plt.xlabel("Input (x)")
plt.ylabel("Output")
plt.grid(True)
plt.axhline(0, linewidth=0.8)
plt.axvline(0, linewidth=0.8)
plt.show()
```

```python
# Comparison of Sigmoid, Tanh and ReLU

x = np.linspace(-10, 10, 400)

y_sigmoid = sigmoid(x)
y_tanh = tanh(x)
y_relu = relu(x)

plt.figure(figsize=(10, 6))

plt.plot(x, y_sigmoid, label="Sigmoid")
plt.plot(x, y_tanh, label="Tanh")
plt.plot(x, y_relu, label="ReLU")

plt.title("Comparison of Activation Functions")
plt.xlabel("Input (x)")
plt.ylabel("Output")
plt.grid(True)
plt.legend()

plt.axhline(0, linewidth=0.8)
plt.axvline(0, linewidth=0.8)

plt.show()
```

```python

# TASK B: PERFORMANCE COMPARISON OF ACTIVATION FUNCTIONS


import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import tensorflow as tf

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Input

print("TensorFlow Version:", tf.__version__)
print("Name: Swathi V")
print("Roll No: 24BAD122")
```

```python
# Load MNIST dataset

(x_train, y_train), (x_test, y_test) = tf.keras.datasets.mnist.load_data()

print("Original Training Data Shape:", x_train.shape)
print("Original Testing Data Shape:", x_test.shape)
```

```python
# Normalize pixel values

x_train = x_train.astype("float32") / 255.0
x_test = x_test.astype("float32") / 255.0

# Flatten 28x28 images into 784 features

x_train = x_train.reshape(-1, 784)
x_test = x_test.reshape(-1, 784)

print("Processed Training Data Shape:", x_train.shape)
print("Processed Testing Data Shape:", x_test.shape)
```

```python
# Function to create identical ANN architectures

def create_model(activation_function):

    model = Sequential([
        Input(shape=(784,)),

        Dense(128, activation=activation_function),
        Dense(64, activation=activation_function),

        Dense(10, activation="softmax")
    ])

    model.compile(
        optimizer="adam",
        loss="sparse_categorical_crossentropy",
        metrics=["accuracy"]
    )

    return model
```

```python
# SIGMOID ANN

sigmoid_model = create_model("sigmoid")

sigmoid_model.summary()

history_sigmoid = sigmoid_model.fit(
    x_train,
    y_train,
    epochs=10,
    batch_size=128,
    validation_split=0.2,
    verbose=1
)
```

```python
# TANH ANN

tanh_model = create_model("tanh")

tanh_model.summary()

history_tanh = tanh_model.fit(
    x_train,
    y_train,
    epochs=10,
    batch_size=128,
    validation_split=0.2,
    verbose=1
)
```

```python
# RELU ANN

relu_model = create_model("relu")

relu_model.summary()

history_relu = relu_model.fit(
    x_train,
    y_train,
    epochs=10,
    batch_size=128,
    validation_split=0.2,
    verbose=1
)
```

```python
# Training Accuracy Comparison

plt.figure(figsize=(10, 6))

plt.plot(
    history_sigmoid.history["accuracy"],
    label="Sigmoid"
)

plt.plot(
    history_tanh.history["accuracy"],
    label="Tanh"
)

plt.plot(
    history_relu.history["accuracy"],
    label="ReLU"
)

plt.title("Training Accuracy Comparison")
plt.xlabel("Epoch")
plt.ylabel("Training Accuracy")
plt.legend()
plt.grid(True)

plt.show()
```

```python
# Validation Accuracy Comparison

plt.figure(figsize=(10, 6))

plt.plot(
    history_sigmoid.history["val_accuracy"],
    label="Sigmoid"
)

plt.plot(
    history_tanh.history["val_accuracy"],
    label="Tanh"
)

plt.plot(
    history_relu.history["val_accuracy"],
    label="ReLU"
)

plt.title("Validation Accuracy Comparison")
plt.xlabel("Epoch")
plt.ylabel("Validation Accuracy")
plt.legend()
plt.grid(True)

plt.show()
```

```python
# Training Loss Comparison

plt.figure(figsize=(10, 6))

plt.plot(
    history_sigmoid.history["loss"],
    label="Sigmoid"
)

plt.plot(
    history_tanh.history["loss"],
    label="Tanh"
)

plt.plot(
    history_relu.history["loss"],
    label="ReLU"
)

plt.title("Training Loss Comparison")
plt.xlabel("Epoch")
plt.ylabel("Training Loss")
plt.legend()
plt.grid(True)

plt.show()
```

```python
# Validation Loss Comparison

plt.figure(figsize=(10, 6))

plt.plot(
    history_sigmoid.history["val_loss"],
    label="Sigmoid"
)

plt.plot(
    history_tanh.history["val_loss"],
    label="Tanh"
)

plt.plot(
    history_relu.history["val_loss"],
    label="ReLU"
)

plt.title("Validation Loss Comparison")
plt.xlabel("Epoch")
plt.ylabel("Validation Loss")
plt.legend()
plt.grid(True)

plt.show()
```

```python
# Combined Accuracy Graph

plt.figure(figsize=(12, 7))

plt.plot(
    history_sigmoid.history["accuracy"],
    label="Sigmoid - Training"
)

plt.plot(
    history_sigmoid.history["val_accuracy"],
    label="Sigmoid - Validation"
)

plt.plot(
    history_tanh.history["accuracy"],
    label="Tanh - Training"
)

plt.plot(
    history_tanh.history["val_accuracy"],
    label="Tanh - Validation"
)

plt.plot(
    history_relu.history["accuracy"],
    label="ReLU - Training"
)

plt.plot(
    history_relu.history["val_accuracy"],
    label="ReLU - Validation"
)

plt.title("Training and Validation Accuracy")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.legend()
plt.grid(True)

plt.show()
```

```python
# Combined Loss Graph

plt.figure(figsize=(12, 7))

plt.plot(
    history_sigmoid.history["loss"],
    label="Sigmoid - Training"
)

plt.plot(
    history_sigmoid.history["val_loss"],
    label="Sigmoid - Validation"
)

plt.plot(
    history_tanh.history["loss"],
    label="Tanh - Training"
)

plt.plot(
    history_tanh.history["val_loss"],
    label="Tanh - Validation"
)

plt.plot(
    history_relu.history["loss"],
    label="ReLU - Training"
)

plt.plot(
    history_relu.history["val_loss"],
    label="ReLU - Validation"
)

plt.title("Training and Validation Loss")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.legend()
plt.grid(True)

plt.show()
```

```python
# Final Accuracy Results

results = {
    "Activation": [
        "Sigmoid",
        "Tanh",
        "ReLU"
    ],

    "Final Training Accuracy": [
        history_sigmoid.history["accuracy"][-1],
        history_tanh.history["accuracy"][-1],
        history_relu.history["accuracy"][-1]
    ],

    "Final Validation Accuracy": [
        history_sigmoid.history["val_accuracy"][-1],
        history_tanh.history["val_accuracy"][-1],
        history_relu.history["val_accuracy"][-1]
    ],

    "Final Training Loss": [
        history_sigmoid.history["loss"][-1],
        history_tanh.history["loss"][-1],
        history_relu.history["loss"][-1]
    ],

    "Final Validation Loss": [
        history_sigmoid.history["val_loss"][-1],
        history_tanh.history["val_loss"][-1],
        history_relu.history["val_loss"][-1]
    ]
}

results_df = pd.DataFrame(results)

display(results_df)
```

```python
# Display accuracy as percentages

display_df = results_df.copy()

display_df["Final Training Accuracy"] = (
    display_df["Final Training Accuracy"] * 100
).round(2)

display_df["Final Validation Accuracy"] = (
    display_df["Final Validation Accuracy"] * 100
).round(2)

display_df["Final Training Loss"] = (
    display_df["Final Training Loss"]
).round(4)

display_df["Final Validation Loss"] = (
    display_df["Final Validation Loss"]
).round(4)

display_df.columns = [
    "Activation",
    "Training Accuracy (%)",
    "Validation Accuracy (%)",
    "Training Loss",
    "Validation Loss"
]

display(display_df)
```

```python
# Find the best activation function

best_index = results_df["Final Validation Accuracy"].idxmax()

best_activation = results_df.loc[
    best_index,
    "Activation"
]

best_accuracy = results_df.loc[
    best_index,
    "Final Validation Accuracy"
]

print("Best Activation Function:", best_activation)
print("Best Validation Accuracy:",
      round(best_accuracy * 100, 2), "%")
```

```python
# Performance Analysis

print("PERFORMANCE ANALYSIS")
print("=" * 50)

for _, row in results_df.iterrows():

    print("\nActivation Function:", row["Activation"])

    print(
        "Training Accuracy:",
        round(row["Final Training Accuracy"] * 100, 2),
        "%"
    )

    print(
        "Validation Accuracy:",
        round(row["Final Validation Accuracy"] * 100, 2),
        "%"
    )

    print(
        "Training Loss:",
        round(row["Final Training Loss"], 4)
    )

    print(
        "Validation Loss:",
        round(row["Final Validation Loss"], 4)
    )

print("\nBest Activation Function:",
      best_activation)

print(
    "Best Validation Accuracy:",
    round(best_accuracy * 100, 2),
    "%"
)
```

```python
# EXPT NO: 2
# TASK C: COMPARISON OF OPTIMIZATION ALGORITHMS
# Name: Swathi V
# Roll No: 24BAD122

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import tensorflow as tf

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Input

print("EXPERIMENT NO: 2")
print("TASK C: COMPARISON OF OPTIMIZATION ALGORITHMS")
print("Name: Swathi V")
print("Roll No: 24BAD122")
print("TensorFlow Version:", tf.__version__)
```

```python
# Load MNIST dataset

(x_train, y_train), (x_test, y_test) = tf.keras.datasets.mnist.load_data()

print("Training data shape:", x_train.shape)
print("Testing data shape:", x_test.shape)
```

```python
# Normalize pixel values

x_train = x_train.astype("float32") / 255.0
x_test = x_test.astype("float32") / 255.0

# Flatten 28 x 28 images into 784 features

x_train = x_train.reshape(-1, 784)
x_test = x_test.reshape(-1, 784)

print("Processed training shape:", x_train.shape)
print("Processed testing shape:", x_test.shape)
```

```python
# Create identical ANN architecture

def create_model(optimizer):

    model = Sequential([
        Input(shape=(784,)),

        Dense(128, activation="relu"),
        Dense(64, activation="relu"),

        Dense(10, activation="softmax")
    ])

    model.compile(
        optimizer=optimizer,
        loss="sparse_categorical_crossentropy",
        metrics=["accuracy"]
    )

    return model
```

```python
# ==========================================
# SGD OPTIMIZER
# ==========================================

sgd_optimizer = tf.keras.optimizers.SGD(
    learning_rate=0.01
)

sgd_model = create_model(sgd_optimizer)

sgd_model.summary()

history_sgd = sgd_model.fit(
    x_train,
    y_train,
    epochs=10,
    batch_size=128,
    validation_split=0.2,
    verbose=1
)
```

```python
# ==========================================
# MOMENTUM OPTIMIZER
momentum_optimizer = tf.keras.optimizers.SGD(
    learning_rate=0.01,
    momentum=0.9
)

momentum_model = create_model(momentum_optimizer)

momentum_model.summary()

history_momentum = momentum_model.fit(
    x_train,
    y_train,
    epochs=10,
    batch_size=128,
    validation_split=0.2,
    verbose=1
)
```

```python
# ==========================================
# RMSPROP OPTIMIZER
# ==========================================

rmsprop_optimizer = tf.keras.optimizers.RMSprop(
    learning_rate=0.001
)

rmsprop_model = create_model(rmsprop_optimizer)

rmsprop_model.summary()

history_rmsprop = rmsprop_model.fit(
    x_train,
    y_train,
    epochs=10,
    batch_size=128,
    validation_split=0.2,
    verbose=1
)
```

```python
# ==========================================
# ADAM OPTIMIZER
# ==========================================

adam_optimizer = tf.keras.optimizers.Adam(
    learning_rate=0.001
)

adam_model = create_model(adam_optimizer)

adam_model.summary()

history_adam = adam_model.fit(
    x_train,
    y_train,
    epochs=10,
    batch_size=128,
    validation_split=0.2,
    verbose=1
)
```

```python
# ==========================================
# LOSS VS EPOCH
# ==========================================

plt.figure(figsize=(12, 7))

plt.plot(
    history_sgd.history["val_loss"],
    label="SGD"
)

plt.plot(
    history_momentum.history["val_loss"],
    label="Momentum"
)

plt.plot(
    history_rmsprop.history["val_loss"],
    label="RMSProp"
)

plt.plot(
    history_adam.history["val_loss"],
    label="Adam"
)

plt.title("Validation Loss vs Epoch")
plt.xlabel("Epoch")
plt.ylabel("Validation Loss")
plt.legend()
plt.grid(True)

plt.show()
```

```python
# ==========================================
# TRAINING LOSS VS EPOCH
# ==========================================

plt.figure(figsize=(12, 7))

plt.plot(
    history_sgd.history["loss"],
    label="SGD"
)

plt.plot(
    history_momentum.history["loss"],
    label="Momentum"
)

plt.plot(
    history_rmsprop.history["loss"],
    label="RMSProp"
)

plt.plot(
    history_adam.history["loss"],
    label="Adam"
)

plt.title("Training Loss vs Epoch")
plt.xlabel("Epoch")
plt.ylabel("Training Loss")
plt.legend()
plt.grid(True)

plt.show()
```

```python
# ==========================================
# ACCURACY VS EPOCH
# ==========================================

plt.figure(figsize=(12, 7))

plt.plot(
    history_sgd.history["val_accuracy"],
    label="SGD"
)

plt.plot(
    history_momentum.history["val_accuracy"],
    label="Momentum"
)

plt.plot(
    history_rmsprop.history["val_accuracy"],
    label="RMSProp"
)

plt.plot(
    history_adam.history["val_accuracy"],
    label="Adam"
)

plt.title("Validation Accuracy vs Epoch")
plt.xlabel("Epoch")
plt.ylabel("Validation Accuracy")
plt.legend()
plt.grid(True)

plt.show()
```

```python
# ==========================================
# TRAINING ACCURACY VS EPOCH
# ==========================================

plt.figure(figsize=(12, 7))

plt.plot(
    history_sgd.history["accuracy"],
    label="SGD"
)

plt.plot(
    history_momentum.history["accuracy"],
    label="Momentum"
)

plt.plot(
    history_rmsprop.history["accuracy"],
    label="RMSProp"
)

plt.plot(
    history_adam.history["accuracy"],
    label="Adam"
)

plt.title("Training Accuracy vs Epoch")
plt.xlabel("Epoch")
plt.ylabel("Training Accuracy")
plt.legend()
plt.grid(True)

plt.show()
```

```python
# ==========================================
# FINAL ACCURACY COMPARISON
# ==========================================

results = {
    "Optimizer": [
        "SGD",
        "Momentum",
        "RMSProp",
        "Adam"
    ],

    "Training Accuracy": [
        history_sgd.history["accuracy"][-1],
        history_momentum.history["accuracy"][-1],
        history_rmsprop.history["accuracy"][-1],
        history_adam.history["accuracy"][-1]
    ],

    "Validation Accuracy": [
        history_sgd.history["val_accuracy"][-1],
        history_momentum.history["val_accuracy"][-1],
        history_rmsprop.history["val_accuracy"][-1],
        history_adam.history["val_accuracy"][-1]
    ],

    "Training Loss": [
        history_sgd.history["loss"][-1],
        history_momentum.history["loss"][-1],
        history_rmsprop.history["loss"][-1],
        history_adam.history["loss"][-1]
    ],

    "Validation Loss": [
        history_sgd.history["val_loss"][-1],
        history_momentum.history["val_loss"][-1],
        history_rmsprop.history["val_loss"][-1],
        history_adam.history["val_loss"][-1]
    ]
}

results_df = pd.DataFrame(results)

display(results_df)
```

```python
# Convert accuracy to percentages

display_df = results_df.copy()

display_df["Training Accuracy"] = (
    display_df["Training Accuracy"] * 100
).round(2)

display_df["Validation Accuracy"] = (
    display_df["Validation Accuracy"] * 100
).round(2)

display_df["Training Loss"] = (
    display_df["Training Loss"]
).round(4)

display_df["Validation Loss"] = (
    display_df["Validation Loss"]
).round(4)

display_df.columns = [
    "Optimizer",
    "Training Accuracy (%)",
    "Validation Accuracy (%)",
    "Training Loss",
    "Validation Loss"
]

display(display_df)
```

```python
# ==========================================
# FIND BEST OPTIMIZER
# ==========================================

best_index = results_df["Validation Accuracy"].idxmax()

best_optimizer = results_df.loc[
    best_index,
    "Optimizer"
]

best_accuracy = results_df.loc[
    best_index,
    "Validation Accuracy"
]

print("Best Optimizer:", best_optimizer)
print(
    "Best Validation Accuracy:",
    round(best_accuracy * 100, 2),
    "%"
)
```

```python
# ==========================================
# CONVERGENCE ANALYSIS
# ==========================================

threshold = 0.95

histories = {
    "SGD": history_sgd,
    "Momentum": history_momentum,
    "RMSProp": history_rmsprop,
    "Adam": history_adam
}

print("CONVERGENCE ANALYSIS")
print("=" * 50)

for name, history in histories.items():

    accuracies = history.history["val_accuracy"]

    convergence_epoch = None

    for epoch, accuracy in enumerate(accuracies, start=1):
        if accuracy >= threshold:
            convergence_epoch = epoch
            break

    if convergence_epoch is not None:
        print(
            name,
            "reached",
            threshold * 100,
            "% validation accuracy at epoch",
            convergence_epoch
        )
    else:
        print(
            name,
            "did not reach",
            threshold * 100,
            "% validation accuracy"
        )
```

```python
# ==========================================
# AUTOMATIC CONVERGENCE SUMMARY
# ==========================================

print("FINAL CONVERGENCE SUMMARY")
print("=" * 50)

for name, history in histories.items():

    final_val_accuracy = history.history["val_accuracy"][-1]
    final_val_loss = history.history["val_loss"][-1]

    print(f"\n{name}")
    print(
        "Final Validation Accuracy:",
        round(final_val_accuracy * 100, 2),
        "%"
    )
    print(
        "Final Validation Loss:",
        round(final_val_loss, 4)
    )

print("\nBest Optimizer:", best_optimizer)
print(
    "Best Validation Accuracy:",
    round(best_accuracy * 100, 2),
    "%"
)
```
